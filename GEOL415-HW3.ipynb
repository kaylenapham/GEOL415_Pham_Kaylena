```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('LA_AQS_2023.csv') 
df.head()
```

```python
list(df)
```

```python
df_o3 = df.loc[df['Parameter Name']=='Ozone']
df_o3_hourly=df_o3.loc[df_o3['Duration Description']=='1 HOUR']
```

```python
df_o3_hourly
```

```python
# Date (Local), AQI
df_o3_hourly['Date (Local)']=pd.to_datetime(df_o3_hourly['Date (Local)'])
```

```python
df_o3_hourly.iloc[0]
```

```python
sns.set_theme(style="white")
plt.figure(figsize=(10, 6), dpi= 300)
sns.lineplot(x='Date (Local)', y='Arithmetic Mean', data=df_o3_hourly, linewidth=2)
plt.xlabel('Date (LST)')
plt.ylabel('Average O3 (ppm)')
plt.title('Daily Average Hourly Ozone')
plt.grid(True)
plt.tight_layout()
plt.show()
```

# Question B

```python
# Question B:

# Converting ppm to ppbv
df_o3_ppbv = df_o3_hourly.copy()
df_o3_ppbv['Arithmetic Mean'] = df_o3_ppbv['Arithmetic Mean']*1000

df_o3_above35 = df_o3_ppbv.loc[df_o3_ppbv['Arithmetic Mean']>35]
df_o3_above35.head()
```

```python
df_o3_above35.iloc[0]
```

```python
days = df_o3_above35['Date (Local)'].dt.date.nunique()

days
```
